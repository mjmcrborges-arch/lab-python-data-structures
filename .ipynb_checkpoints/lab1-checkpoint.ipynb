```python
# 1. Create the list of products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# 2. Create an empty inventory dictionary
inventory = {}

# 3. Ask the user for the quantity of each product
for product in products:
    quantity = int(input(f"How many {product}s are available? "))
    inventory[product] = quantity
    print (inventory)

# 4. Create an empty set for customer orders
customer_orders = set()

# 5. Ask the user for three products to order
for i in range(3):
    product = input(f"Enter product {i + 1} to order: ")
    
    if product in products:
        customer_orders.add(product)
    else:
        print("Product not found. Please choose from:", products)

# 6. Print the products in the customer_orders set
print("Customer orders:", customer_orders)
# 7. Calculate order statistics
total_products_ordered = len(customer_orders)
unique_products_ordered = len(customer_orders)

percentage_unique_products_ordered = (
    unique_products_ordered / len(products)
) * 100

percentage_products_ordered = (
    total_products_ordered / len(products)
) * 100

# Store the statistics in a tuple
order_status = (
    total_products_ordered,
    unique_products_ordered,
    percentage_unique_products_ordered,
    percentage_products_ordered
)

# 8. Print the order statistics
print("\nOrder Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Unique Products Ordered: {order_status[1]}")
print(f"Percentage of Unique Products Ordered: {order_status[2]:.2f}%")
print(f"Percentage of Products Ordered: {order_status[3]:.2f}%")

# 9. Update the inventory by subtracting 1
for product in customer_orders:
    inventory[product] -= 1

# Display the updated inventory
print("\nUpdated Inventory:")
print(inventory)











```
